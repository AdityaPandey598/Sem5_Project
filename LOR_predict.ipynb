```python
import random
from faker import Faker
from openai import OpenAI
import time
fake = Faker()
apki_key="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"
    
def generate_LOR():
    client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=apki_key,
    )
    name=fake.name()
    relationship=random.choice(['professor', 'employer', 'mentor', 'colleague']),
    nature=random.choice(["High Positivity, High Specificity, High Relevance",
                         "Moderate Positivity, Medium Specificity, Medium Relevance",
                         "Low Positivity, Low Specificity, Low Relevance",
                         "High Positivity, Low Specificity, Medium Relevance",
                         "Medium Positivity, High Specificity, Low Relevance"])
    prompt = f'''
    You are a {relationship} writing a letter of recommendation of {name}. The nature of the letter is {nature}.
    Write a concise letter of recommendation (100 words) for {name} that reflects the specified nature. Ensure the letter includes specific examples and relevant achievements that align with the chosen positivity, specificity, and relevance levels.
    '''

    completion = client.chat.completions.create(
    model="deepseek/deepseek-chat-v3.1:free",
    messages=[{"role": "user","content": prompt}]
                                                )
    return completion.choices[0].message.content




for _ in range(100):
    lor = []
    with open('my_data.txt', 'a') as file:
        for i in range(1):
            print(f"Generating LOR {i + 1}")
            lor.append(generate_LOR())
            print("Waiting before next request...")
            file.write(lor[-1] + "\n")
            time.sleep(5)




```
